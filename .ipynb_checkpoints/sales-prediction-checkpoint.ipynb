```python
# Install dependencies as needed:
# pip install kagglehub[pandas-datasets]
import kagglehub
import os
import pandas as pd
from kagglehub import KaggleDatasetAdapter

# Create a writable directory for cache if needed
# Use a directory you have write access to
os.environ['KAGGLEHUB_CACHE_DIR'] = './kaggle_cache'  # Set cache to a writable location

try:
    # Try to load using kagglehub
    df = kagglehub.dataset_load(
        KaggleDatasetAdapter.PANDAS,
        "yasserh/walmart-dataset",
        "Walmart.csv"  # Simplified path - just the filename
    )
except Exception as e:
    print(f"Error with kagglehub: {e}")
    print("Falling back to direct pandas read if the file is already available...")
    # Alternative: If the dataset is already available in the input directory
    try:
        df = pd.read_csv("/kaggle/input/yasserh/walmart-dataset/Walmart.csv")
    except:
        # If that fails too, try the original path
        df = pd.read_csv("/kaggle/input/datasets/yasserh/walmart-dataset/Walmart.csv")

print("First 5 records:", df.head())
```

```python
df['Date'] = pd.to_datetime(df['Date'],format ='%d-%m-%Y')
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Week'] = df['Date'].dt.isocalendar().week
```

```python
df.drop(columns='Date',axis=1,inplace=True)
```

```python
df.shape
```

```python
df = df.sort_values(['Store', 'Year', 'Week'])
df.head()
```

```python
df['lag_1'] = df.groupby('Store')['Weekly_Sales'].shift(1)
df['lag_4'] = df.groupby('Store')['Weekly_Sales'].shift(4)
df['lag_12'] = df.groupby('Store')['Weekly_Sales'].shift(12)
df['lag_52'] = df.groupby('Store')['Weekly_Sales'].shift(52)
```

```python
df.head()
```

```python
df['rolling_mean_4'] = df.groupby('Store')['Weekly_Sales'].transform(lambda x: x.shift(1).rolling(4).mean())
df['rolling_mean_12'] = df.groupby('Store')['Weekly_Sales'].transform(lambda x: x.shift(1).rolling(12).mean())
```

```python
df.head()
```

```python
df = df.dropna()
```

# Splitting Data

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
train = df[df['Year'] < 2012]
test = df[df['Year'] >= 2012]
```

```python
features = [
    'Store', 'Holiday_Flag', 'Temperature', 'Fuel_Price',
    'CPI', 'Unemployment', 'Year', 'Month', 'Week',
    'lag_1', 'lag_4', 'lag_12', 'lag_52',
    'rolling_mean_4', 'rolling_mean_12'
]

X_train = train[features]
y_train = train['Weekly_Sales']

X_test = test[features]
y_test = test['Weekly_Sales']
```

```python
df.info()
```

# Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_absolute_error,root_mean_squared_error
linear_model = LinearRegression()
linear_model.fit(X_train,y_train)

y_predlm= linear_model.predict(X_test)
print("R2 score :",r2_score(y_test,y_predlm))
print("MAE score :",mean_absolute_error(y_test,y_predlm))
print("RMSE score :",root_mean_squared_error(y_test,y_predlm))
```

# Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
RFG_model = RandomForestRegressor(n_estimators=500,max_depth=20,min_samples_split=5
                                 ,min_samples_leaf=2,max_features='sqrt',n_jobs=-1)
RFG_model.fit(X_train,y_train)

y_predRFG = RFG_model.predict(X_test)
y_predRFG_train = RFG_model.predict(X_train)
print("____________________test_______________________")
print("R2 score :",r2_score(y_test,y_predRFG))
print("MAE score :",mean_absolute_error(y_test,y_predRFG))
print("RMSE score :",root_mean_squared_error(y_test,y_predRFG))
print("___________________train_________________________")
print("R2 score :",r2_score(y_train,y_predRFG_train))
print("MAE score :",mean_absolute_error(y_train,y_predRFG_train))
print("RMSE score :",root_mean_squared_error(y_train,y_predRFG_train))
```

# XgBOOST

```python
from xgboost import XGBRegressor

XGB_model = XGBRegressor(
    n_estimators=800,
    learning_rate=0.03,
    max_depth=6,              # reduce depth
    min_child_weight=5,       # increase
    subsample=0.7,
    colsample_bytree=0.7,
    reg_alpha=0.5,            # add regularization
    reg_lambda=2,
    random_state=42
)
XGB_model.fit(X_train,y_train)

y_predXGB = XGB_model.predict(X_test)
y_predXGB_train = XGB_model.predict(X_train)
print("____________________test_______________________")
print("R2 score :",r2_score(y_test,y_predXGB))
print("MAE score :",mean_absolute_error(y_test,y_predXGB))
print("RMSE score :",root_mean_squared_error(y_test,y_predXGB))
print("___________________train_________________________")
print("R2 score :",r2_score(y_train,y_predXGB_train))
print("MAE score :",mean_absolute_error(y_train,y_predXGB_train))
print("RMSE score :",root_mean_squared_error(y_train,y_predXGB_train))
```

## Hyperparameter tuning for XgBoost

```python
from sklearn.model_selection import TimeSeriesSplit, RandomizedSearchCV
from xgboost import XGBRegressor

tscv = TimeSeriesSplit(n_splits=5)

param_grid = {
    'n_estimators': [300, 500, 800],
    'learning_rate': [0.01, 0.03, 0.05],
    'max_depth': [4, 6, 8],
    'min_child_weight': [1, 3, 5],
    'subsample': [0.7, 0.8],
    'colsample_bytree': [0.7, 0.8],
    'reg_alpha': [0, 0.1, 0.5],
    'reg_lambda': [1, 2]
}

xgb = XGBRegressor(random_state=42)

search = RandomizedSearchCV(
    xgb,
    param_distributions=param_grid,
    n_iter=20,
    cv=tscv,
    scoring='neg_mean_absolute_error',
    verbose=1,
    n_jobs=-1
)

search.fit(X_train, y_train)

best_model = search.best_estimator_
```

```python
# Predictions
y_pred_best = best_model.predict(X_test)
y_pred_best_train = best_model.predict(X_train)

print("____________________test_______________________")
print("R2 score :", r2_score(y_test, y_pred_best))
print("MAE score :", mean_absolute_error(y_test, y_pred_best))
print("RMSE score :", root_mean_squared_error(y_test, y_pred_best))

print("___________________train_________________________")
print("R2 score :", r2_score(y_train, y_pred_best_train))
print("MAE score :", mean_absolute_error(y_train, y_pred_best_train))
print("RMSE score :", root_mean_squared_error(y_train, y_pred_best_train))
```

```python
print("Best Parameters:", search.best_params_)
```
